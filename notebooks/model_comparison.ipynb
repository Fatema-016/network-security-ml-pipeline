```python
import matplotlib.pyplot as plt
import numpy as np

models = ['Logistic\nRegression', 'Random\nForest', 'XGBoost']
test_f1       = [0.9110, 0.9824, 0.9830]
test_recall   = [0.9248, 0.9926, 0.9903]
test_precision= [0.8976, 0.9725, 0.9759]
test_roc_auc  = [0.9338, 0.9887, 0.9885]

x     = np.arange(len(models))
width = 0.2

fig, ax = plt.subplots(figsize=(12, 6))

bars1 = ax.bar(x - 1.5*width, test_f1,        width, label='F1 Score',  color='#2196F3')
bars2 = ax.bar(x - 0.5*width, test_recall,    width, label='Recall',    color='#4CAF50')
bars3 = ax.bar(x + 0.5*width, test_precision, width, label='Precision', color='#FF9800')
bars4 = ax.bar(x + 1.5*width, test_roc_auc,   width, label='ROC-AUC',   color='#9C27B0')

ax.set_xlabel('Model', fontsize=12)
ax.set_ylabel('Score', fontsize=12)
ax.set_title(
    'Model Comparison — CICIDS 2017 Network Security\n'
    'Binary Classification: BENIGN vs ATTACK',
    fontsize=13,
    fontweight='bold'
)
ax.set_xticks(x)
ax.set_xticklabels(models, fontsize=11)
ax.set_ylim(0.85, 1.02)
ax.legend(fontsize=10)
ax.grid(axis='y', alpha=0.3)


for bar in [bars1, bars2, bars3, bars4]:
    for rect in bar:
        height = rect.get_height()
        ax.annotate(
            f'{height:.4f}',
            xy=(rect.get_x() + rect.get_width()/2, height),
            xytext=(0, 3),
            textcoords="offset points",
            ha='center',
            va='bottom',
            fontsize=7
        )

plt.tight_layout()
plt.savefig('C:/Users/alefi/Documents/Courses/Complete_MLOps_Course_KN/network_security_project/screenshots', dpi=150, bbox_inches='tight')
plt.show()
print("Chart saved to screenshots/model_comparison_bar.png")
```
